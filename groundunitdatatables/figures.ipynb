```python
for counter in [False, True]:

    startgamesetup(
        None,
        sheets=[["P"]],
        style="temperate",
        writelogfiles=False,
        dotsperhex=120,
    )

    x = 3
    y = 3

    for s in [

        "infantry platoon",
        "infantry HQ platoon",
        "infantry FAC section",
        "",
        "",
        "",

        "heavy tank platoon",
        "medium tank platoon",
        "light tank platoon",
        "heavy tank HQ platoon",
        "medium tank HQ platoon",
        "light tank HQ platoon",

        "medium IFV platoon",
        "light IFV platoon",
        "heavy APC platoon",
        "medium APC platoon",
        "light APC platoon",
        "",


        "truck platoon",
        "light truck platoon",
        "light truck section",
        "",
        "",
        "",

        "towed artillery battery",
        "armored artillery battery",
        "mobile artillery battery",
        "armored rocket artillery battery",
        "mobile rocket artillery battery",
        "mobile missile artillery battery",

    ]:
        if s != "":
            if s.startswith("KS"):
                azimuth = "ENE"
            else:
                azimuth = None
            hexcode = "P-%02d%02d" % (x + 0, y)
            name = s
            name = name.replace(" section", " ")
            name = name.replace(" platoon", "  ")
            name = name.replace(" battery", "   ")
            setupgroundunit(name, hexcode, s, azimuth=azimuth, color="lightgreen", counter=counter, identified=True)
        if y == 8:
            x += 2
            y = 3
        else:
            y += 1

    drawmap(zoom=False,xmin=1.5,xmax=13.5,ymin=16.5,ymax=23.5)

!mv map-00.png figure-ground-combat-units.png
```
