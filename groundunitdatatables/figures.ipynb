```python
from glass import *
```

```python
startgamesetup(
    None,
    sheets=[["P"]],
    style="temperate",
    writelogfiles=False,
    dotsperhex=120,
)

x = 3
y = 3

for s in [

    "infantry",
    "infantry HQ",
    "infantry FAC",
    "",
    "",
    "",

    "heavy tank",
    "medium tank",
    "light tank",
    "heavy tank HQ",
    "medium tank HQ",
    "light tank HQ",

    "medium IFV",
    "light IFV",
    "heavy APC",
    "medium APC",
    "light APC",
    "",


    "truck",
    "light truck",
    "",
    "",
    "",
    "",

    "towed artillery",
    "armored artillery",
    "mobile artillery",
    "armored rocket artillery",
    "mobile rocket artillery",
    "mobile missile artillery",

    "armored CCU",
    "mobile CCU",
    "towed EWR-A",
    "mobile EWR-A",
    "towed FCR-B",
    "mobile FCR-B",
]:
    if s != "":
        if s.startswith("KS"):
            azimuth = "ENE"
        else:
            azimuth = None
        hexcode = "P-%02d%02d" % (x + 0, y)
        name = s.replace(" ", "\n  ")
        setupgroundunit(name, hexcode, s, azimuth=azimuth, color="lightgreen", identified=True)
    if y == 8:
        x += 2
        y = 3
    else:
        y += 1

drawmap(zoom=False,xmin=1.5,xmax=14.5,ymin=16.5,ymax=23.5)
```
