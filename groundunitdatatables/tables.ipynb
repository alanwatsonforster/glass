```python
import glass.jsonc

latexdir = "./"
jsondir = "../src/glass/groundunitdata/"
```

```python
def prettyname(name):
    name = name.title()
    for word in [
        "IFV",
        "APC",
        "HQ",
        "FAC",
        "SAM",
        "DShK",
        "ZU",
        "ZPU",
        "ZSU",
        "GDF",
        "BOFI",
        "KS",
        "DCA",
        "AMX",
        "TCM",
        "LAV",
        "AD",
        "SA",
    ]:
        name = name.replace(word.title(), word)
    for word in [
        "squad",
        "section",
        "platoon",
        "battery",
    ]:
        name = name.replace(word.title(), "")
    return name


def echelon(symbols):
    if "squad" in symbols:
        return "Squad"
    elif "section" in symbols:
        return "Section"
    elif "platoon" in symbols:
        return "Platoon"
    elif "company" in symbols:
        return "Company"
    elif "battery" in symbols:
        return "Battery"
    else:
        return ""


def VPtext(VP):
    if VP is None:
        return r"\wbox[c]{0}{--}"
    else:
        return VP
```

```python
with open(latexdir + "/" + "table-ground-combat-units.tex", "w") as latexfile:

    def writelatex(s=""):
        print(s, file=latexfile, end="")

    def writelatexline(s=""):
        print(s, file=latexfile, end="\n")

    writelatexline(r"\begin{tabularx}{0.5\linewidth}{Lccccc}")
    writelatexline(r"\toprule")
    writelatexline(
        r"""
            Type&
            \vertical{Echelon}&
            \vertical{\begin{tabular}{@{}l@{}}Defense\\Strength\end{tabular}}&
            \vertical{\begin{tabular}{@{}l@{}}Sighting\\Range\end{tabular}}&
            \vertical{\begin{tabular}{@{}l@{}}VPs\\K/2D/D\end{tabular}}&
            \vertical{\begin{tabular}{@{}l@{}}AAA\\Class\tablenotemark{1}\end{tabular}}
            \\
        """
    )
    writelatexline(r"\midrule")
    writelatexline(r"\addlinespace")

    for name in [
        "infantry platoon",
        "infantry HQ platoon",
        "infantry FAC section",
        "",
        "heavy tank platoon",
        "heavy tank HQ platoon",
        "",
        "medium tank platoon",
        "medium tank HQ platoon",
        "",
        "light tank platoon",
        "light tank HQ platoon",
        "",
        "medium IFV platoon",
        "light IFV platoon",
        "",
        "heavy APC platoon",
        "medium APC platoon",
        "light APC platoon",
        "",
        "truck platoon",
        "light truck platoon",
        "light truck section",
        "light truck squad",
        "",
        "towed artillery battery",
        "",
        "armored artillery battery",
        "armored rocket artillery battery",
        "",
        "mobile artillery battery",
        "mobile rocket artillery battery",
        "mobile missile artillery battery",
    ]:

        if name == "":
            writelatexline(r"\addlinespace")
            continue

        print(name)
        data = glass.jsonc.load(open(jsondir + "/" + name + ".json"))


        defensestrength = data["defensestrength"]
        sightingrange = data["sightingrange"]
        VPs = data["VPs"]

        writelatex(r"%s" % prettyname(name))
        writelatex(r"&\wbox[l]{Section}{%s}" % echelon(data["symbols"]))
        writelatex(r"&\wbox[l]{0H}{%s}" % defensestrength)
        writelatex(r"&\wbox{00}{%d}" % sightingrange)
        if "aaa" in data:
            aaa = data["aaa"]
            aaaclass = aaa["class"]
            aaaaltitude = aaa["maximumrelativealtitude"]
            aaaclass = aaa["class"] + str(aaa["maximumrelativealtitude"])
        else:
            aaaclass = "---"
        writelatex(r"&\wbox{00}{%s}/%s/%s" % (VPtext(VPs[0]), VPtext(VPs[1]), VPtext(VPs[2])))
        writelatex(r"&%s" % aaaclass)
        writelatex(r"\\")
        writelatexline()

    writelatexline(r"\addlinespace")
    writelatexline(r"\bottomrule")
    writelatexline(r"\end{tabularx}")
    writelatexline(r"\begin{tablenote}{0.5\linewidth}")
    writelatexline(
        r"\tablenotemark{1}~B2 or B3 indicate that the unit is capable of barrage fire to an altitude of 2 or 3 levels, respectively. Otherwise the unit is not capable of AAA fire."
    )
    writelatexline(r"\end{tablenote}")
```

```python
with open(latexdir + "/" + "table-aaa-units.tex", "w") as latexfile:

    def writelatex(s=""):
        print(s, file=latexfile, end="")

    def writelatexline(s=""):
        print(s, file=latexfile, end="\n")

    writelatexline(r"\begin{tabularx}{\linewidth}{Llccccccc@{~}c@{~}cc@{~}c@{~}ccccl}")
    writelatexline(r"\toprule")
    writelatexline(r"\\")
    writelatexline(
        r"""
            &
            &
            &
            &
            &
            &
            &
            &
            \multicolumn{3}{c}{Range}   &
            \multicolumn{3}{c}{Hit}     &
            &
            &
            &
            \\
        """
    )
    writelatexline(
        r"""
            \cmidrule(lr){9-11}
            \cmidrule(lr){12-14}
        """
    )
    writelatexline(
        r"""
            Type&
            \multicolumn{1}{c}{\multirow[b]{-3}{*}{\vertical{Gun}}}&
            \vertical{Echelon}&
            \multirow[b]{-3}{*}{\vertical{\begin{tabular}{@{}l@{}}Defense\\Strength\end{tabular}}}&
            \multirow[b]{-3}{*}{\vertical{\begin{tabular}{@{}l@{}}Sighting\\Range\end{tabular}}}&
            \vertical{\begin{tabular}{@{}l@{}}VPs\\K/2D/D\end{tabular}}&
            \vertical{\begin{tabular}{@{}l@{}}AAA\\Class\end{tabular}}&
            \multirow[b]{-3}{*}{\vertical{Altitude}}&
            S&M&L&       
            S&M&L&
            \multirow[b]{-3}{*}{\vertical{\begin{tabular}{@{}l@{}}Damage\\Rating\end{tabular}}}&
            \multirow[b]{-3}{*}{\vertical{FCR}}&
            \multirow[b]{-3}{*}{\vertical{\begin{tabular}{@{}l@{}}Night IR\\Sights\end{tabular}}}&
            \multirow[b]{-3}{*}{\vertical{SAM}}\\
        """
    )
    writelatexline(r"\midrule")
    writelatexline(r"\addlinespace")


    for name in [
        "M2 platoon",
        "DShK-38 platoon",
        "M16 section",
        "M55 battery",
        # "mobile M55 section",
        "",
        "ZPU-1 battery",
        "ZPU-2 battery",
        "ZPU-4 battery",
        "mobile ZPU-4 section",
        "",
        "TCM-20 battery",
        "M3 TCM-20 section",
        "Rh-202 battery",
        "Panhard M3 DCA section",
        "M163 section",
        "M167 battery",
        "",
        "ZU-23 battery",
        "mobile ZU-23 section",
        "ZSU-23-4 section",
        "",
        "LAV-AD section",
        "",
        "M53/59 section",
        "AMX-30SA section",
        "Tunguska section",
        "Tunguska-M section",
        # "Pantsir section",
        "",
        "Oerlikon GDF battery",
        "Gepard section",
        "",
        "M-38 battery",
        "",
        "Bofors L/60 battery",
        "Bofors L/70 battery",
        "Bofors L/70 BOFI battery",
        "Bofors L/70 BOFI-R battery",
        "",
        "S-60 battery",
        "",
        "KS-12 battery",
        "",
        "KS-19 battery",
    ]:

        if name == "":
            writelatexline(r"\addlinespace")
            continue

        print(name)
        data = glass.jsonc.load(open(jsondir + "/" + name.replace("/", "") + ".json"))

        defensestrength = data["defensestrength"]
        sightingrange = data["sightingrange"]
        if "VPs" in data:
            VPs = data["VPs"]
        else:
            VPs = [0, 0, 0]

        aaa = data["aaa"]

        aaatype = aaa["type"]
        if "×" in aaatype:
            aaatype = r"\binarymultiply{%s mm}{%s}" % tuple(aaatype.split("×"))
        else:
            aaatype = r"%s mm" % aaatype

        if "sam" in data:
            samtype = data["sam"]["type"]
        else:
            samtype = "---"

        aaaclass = aaa["class"]
        aaarange = aaa["range"]
        aaahitroll = aaa["hitroll"]
        aaaaltitude = aaa["maximumrelativealtitude"]
        aaadamagerating = aaa["damagerating"]

        if "fcrclass" not in aaa:
            aaafcr = "---"
        else:
            aaafcr = aaa["fcrclass"] + "/" + aaa["fcrfrequency"]

        if "nightirsights" not in aaa:
            aaanightirsights = "---"
        else:
            aaanightirsights = "Y"

        writelatex(r"%s" % prettyname(name))
        writelatex(r"&%s" % aaatype)

        writelatex(r"&\wbox[l]{Section}{%s}" % echelon(data["symbols"]))
        writelatex(r"&\wbox[l]{0H}{%s}" % defensestrength)
        writelatex(r"&\wbox{00}{%d}" % sightingrange)
        writelatex(
            r"&\wbox{00/00/0}{%s/%s/%s}" % (VPtext(VPs[0]), VPtext(VPs[1]), VPtext(VPs[2]))
        )

        writelatex(r"&%s" % aaaclass)


        writelatex(r"&\wbox{00}{%s}" % aaaaltitude)

        writelatex(r"&\wbox{00}{%s}" % aaarange[0])
        writelatex(r"&\wbox{00}{%s}" % aaarange[1])
        writelatex(r"&\wbox{00}{%s}" % aaarange[2])

        writelatex(r"&\wbox{00}{%s}" % aaahitroll[0])
        writelatex(r"&\wbox{00}{%s}" % aaahitroll[1])
        writelatex(r"&\wbox{00}{%s}" % aaahitroll[2])

        writelatex(r"&\wbox{0}{%s}" % aaadamagerating)
        writelatex(r"&%s" % aaafcr)
        writelatex(r"&%s" % aaanightirsights)
        writelatex(r"&%s" % samtype)

        writelatex(r"\\")
        writelatexline()

    writelatexline(r"\addlinespace")
    writelatexline(r"\bottomrule")
    writelatexline(r"\end{tabularx}")
```

```python
with open(latexdir + "/" + "table-sam-launcher-units.tex", "w") as latexfile:

    def writelatex(s=""):
        print(s, file=latexfile, end="")

    def writelatexline(s=""):
        print(s, file=latexfile, end="\n")

    writelatexline(r"\begin{tabularx}{\linewidth}{Lcclcccccccccccl}")
    writelatexline(r"\toprule")
    writelatexline(
        r"""
            &
            &
            &
            &
            \multicolumn{3}{c}{EWR}         &
            \multicolumn{2}{c}{TTR}         &
            &
            \multicolumn{2}{c}{Missiles}    &
            &
            \multicolumn{2}{c}{Lock-On}     &
            \\
        """
    )

    writelatexline(
        r"""
            \cmidrule(lr){5-7}
            \cmidrule(lr){8-9}
            \cmidrule(lr){11-12}
            \cmidrule(lr){14-15}
        """
    )
    writelatexline(
        r"""
            Name&
            \vertical{\begin{tabular}{@{}l@{}}Defense\\Strength\end{tabular}}&
            \vertical{\begin{tabular}{@{}l@{}}Sighting\\Range\end{tabular}}&
            \vertical{SAM}          &
            \vertical{Frequency}    &
            \vertical{Range}        &
            \vertical{MTI}          &
            \vertical{Frequency}    &
            \vertical{Range}        &
            \vertical{\begin{tabular}{@{}l@{}}Multi-Target\\Capability\end{tabular}}&
            \vertical{Ready}        &
            \vertical{Volley}       &
            \vertical{\begin{tabular}{@{}l@{}}Quick\\Reaction\end{tabular}}&
            \vertical{Radar}        &
            \vertical{Optical}      &
            \vertical{Gun}          \\
            """
    )
    writelatexline(r"\midrule")

    i = 0

    for name in [
        "SA-2B battery",
        "SA-2C battery",
        "SA-2E battery",
        "SA-2F battery",
        "SA-3A battery",
        "SA-3B section",
        "SA-4 battery",
        "SA-5 battery",
        "SA-6 battery",
        "SA-8A squad",
        "SA-8B squad",
        "SA-9 squad",
        "SA-10A battery",
        "SA-10B battery",
        "SA-11 squad",
        "SA-12A battery",
        "SA-12B battery",
        "SA-13 squad",
        "SA-15 squad",
        "Tunguska section",
        "Pantsir S1 section",
        "SA-17 squad",
        "Pantsir S1M section",
    ]:

        if i % 3 == 0:
            writelatexline(r"\addlinespace")
        i += 1

        print(name)
        data = glass.jsonc.load(open(jsondir + "/" + name + ".json"))

        defensestrength = data["defensestrength"]
        sightingrange = data["sightingrange"]

        sam = data["sam"]

        samtype = sam["type"]

        if "aaa" not in data:
            aaatype = "---"
        else:
            aaatype = data["aaa"]["type"]
            if "×" in aaatype:
                aaatype = r"\binarymultiply{%s mm}{%s}" % tuple(aaatype.split("×"))
            else:
                aaatype = r"%s mm" % aaatype

        if "ewrfrequency" not in sam:
            ewrfrequency = "---"
            ewrrange = "---"
        else:
            ewrfrequency = sam["ewrfrequency"]
            ewrrange = sam["ewrrange"]
        if "ewrmti" not in sam or not sam["ewrmti"]:
            ewrmti = "---"
        else:
            ewrmti = "Y"
        if "ttrfrequency" not in sam:
            ttrfrequency = "---"
            ttrrange = "---"
        else:
            ttrfrequency = sam["ttrfrequency"]
            ttrrange = sam["ttrrange"]

        if "multitargetcapability" not in sam:
            multitargetcapability = "---"
        else:
            multitargetcapability = sam["multitargetcapability"]

        readymissiles = sam["readymissiles"]
        volleymissiles = sam["volleymissiles"]

        if "quickreaction" not in sam or not sam["quickreaction"]:
            quickreaction = "---"
        else:
            quickreaction = "Y"

        if "radarlockon" not in sam:
            radarlockon = "---"
        else:
            radarlockon = sam["radarlockon"]
        if "opticallockon" not in sam:
            opticallockon = "---"
        else:
            opticallockon = sam["opticallockon"]

        writelatex(r"%s" % name)

        writelatex(r"&\wbox[l]{0H}{%s}" % defensestrength)
        writelatex(r"&\wbox{00}{%d}" % sightingrange)

        writelatex(r"&%s" % samtype)
        writelatex(r"&%s" % ewrfrequency)
        writelatex(r"&\wbox{000}{%s}" % ewrrange)
        writelatex(r"&%s" % ewrmti)
        writelatex(r"&%s" % ttrfrequency)
        writelatex(r"&\wbox{00}{%s}" % ttrrange)
        writelatex(r"&%s" % multitargetcapability)
        writelatex(r"&\wbox{00}{%d}" % readymissiles)
        writelatex(r"&\wbox{00}{%d}" % volleymissiles)
        writelatex(r"&%s" % quickreaction)
        writelatex(r"&%s" % radarlockon)
        writelatex(r"&%s" % opticallockon)

        writelatex(r"&%s" % aaatype)

        writelatex(r"\\")
        writelatexline()

    writelatexline(r"\addlinespace")
    writelatexline(r"\bottomrule")
    writelatexline(r"\end{tabularx}")
```
