```python
startgamesetup(None, sheets=[["F"]], variants=["use version 2.4 rules","use house rules"], style="temperate", dotsperhex=300, writelogfiles=False)
startgamesetup(None, sheets=[["F"]], variants=["use version 2.4 rules","use house rules"], style="arid", writelogfiles=False)

import glass.draw
import glass.hexcode


x = 8
y = 3
for c in [ "white", "natoblue", "natored", "natogreen", "natoyellow", "green", "lightgreen", "olivedrab", "tan", "lighttan", "sand", "lightsand", "khaki","airstrikegreen", "airstrikelightgreen", "airstriketan" ]:
    hexcode = "F-%02d%02d" % (x, y)
    setupgroundunit(c, hexcode, "infantry", color=c, identified=True)
    y += 1
    if y == 8:
        x += 1
        y = 3

x = 10
y = 9
hexcode = "F-%02d%02d" % (x, y)
setupgroundunit("1/1", hexcode, "infantry", stack="1/1", color="lightgreen", identified=True)
y += 1
hexcode = "F-%02d%02d" % (x, y)
setupgroundunit("1/2", hexcode, "infantry", stack="1/2", color="lightgreen", identified=True)
setupgroundunit("2/2", hexcode, "infantry", stack="2/2", color="lightgreen", identified=True)
x += 1
y -= 1
hexcode = "F-%02d%02d" % (x, y)
setupgroundunit("1/3", hexcode, "infantry", stack="1/3", color="lightgreen", identified=True)
setupgroundunit("2/3", hexcode, "infantry", stack="2/3", color="lightgreen", identified=True)
setupgroundunit("3/3", hexcode, "infantry", stack="3/3", color="lightgreen", identified=True)
y += 1
hexcode = "F-%02d%02d" % (x, y)
setupgroundunit("1/4", hexcode, "infantry", stack="1/4", color="lightgreen", identified=True)
setupgroundunit("2/4", hexcode, "infantry", stack="2/4", color="lightgreen", identified=True)
setupgroundunit("3/4", hexcode, "infantry", stack="3/4", color="lightgreen", identified=True)
setupgroundunit("4/4", hexcode, "infantry", stack="4/4", color="lightgreen", identified=True)



x = 8
y = 9
hexcode = "F-%02d%02d" % (x, y)
setupgroundunit("entrenched", hexcode, "infantry", stack="1/1", color="lightgreen", identified=True, protectionclass="entrenched")
y += 1
hexcode = "F-%02d%02d" % (x, y)
setupgroundunit("bunkered", hexcode, "infantry", stack="1/1", color="lightgreen", identified=True, protectionclass="bunkered")


x = 13
y = 3
for s in [ 
    "infantry",
    "infantry HQ",
    "medium tank",
    "truck",
    "mobile EWR-A",
    "mobile FCR-A",
    "mobile CCU",
    "SA-2B", "SA-2C", "SA-2E", "SA-2F",
    "KS-12", "KS-19", "S-60", "M-38", "ZPU-4", "mobile ZPU-4", "ZPU-2", "ZPU-1", "ZU-23", "mobile ZU-23", "ZSU-57-2", "ZSU-23-4", "Tunguska", "Pantsir S1", "Pantsir S1M", 
    "M55", "mobile M55", "M16", "M42",
    ]:
    hexcode = "F-%02d%02d" % (x, y)
    if s.startswith("KS"):
        azimuth = "ENE"
    else:
        azimuth = None
    setupgroundunit(s + " ", hexcode, s, azimuth=azimuth, color="lightgreen", identified=True)
    y += 1
    if y == 12:
        x += 1
        y = 3

endgamesetup()
drawmap(zoom=False, xmin=2, xmax=18, ymin=13, ymax=23.5)
#drawmap(zoom=False, xmin=2, xmax=16, ymin=13, ymax=23.5, compactstacks=False)

x = 3
y = 3
for s in [
    "air-defense", "ammunition", "antiarmor", "armor", "artillery", "barge", "bridge", "building", "fixedwing", "fuel", "gun", "hangar",
    "headquarters", "heavy", "infantry", "infantrycarrier", "junk", "largebuilding", "light", "locomotive", "medium", "missile", "ordnance", "radar", "railcar", "reconnaissance", "rocket", "rotarywing", "shelter", "supply", "tower", "transportation", "truck", "unidentified"
    ]:
    hexcode = "F-%02d%02d" % (x, y)
    glass.draw.drawgroundunit(*glass.hexcode.toxy(hexcode), None, [s], "", "", None, s, "", "lightgreen")
    y += 1
    if y == 12:
        x += 1
        y = 3
drawmap(zoom=False, xmin=2, xmax=18, ymin=13, ymax=23.5)

```
